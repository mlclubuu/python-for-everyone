## ১. সিদ্ধান্ত গ্রহণ বা ডিসিশন মেকিং (if / elif / else)
প্রোগ্রামিংয়ে বাস্তব জীবনের মতো বিভিন্ন শর্ত বা কন্ডিশনের ওপর ভিত্তি করে সিদ্ধান্ত নিতে হয়। পাইথনে এর জন্য `if`, `elif` (else if), এবং `else` ব্যবহার করা হয়।

**নিয়ম (Syntax):**
- `if` এর পর শর্ত দিতে হয়। শর্ত সত্যি (True) হলে এর ভেতরের কোড রান করবে।
- একাধিক শর্ত থাকলে `elif` ব্যবহার করতে হয়।
- কোনো শর্তই সত্যি না হলে সবশেষে `else` এর কোড রান করে।
- পাইথনে ব্লক বোঝানোর জন্য **ইন্ডেন্টেশন (Indentation বা ফাঁকা জায়গা)** খুবই গুরুত্বপূর্ণ।

```python
# if, elif, else এর উদাহরণ
age = int(input("আপনার বয়স লিখুন: "))

if age >= 18:
    print("আপনি ভোটার হওয়ার যোগ্য।")
elif age >= 13:
    print("আপনি একজন টিনএজার।")
else:
    print("আপনি এখনো শিশু।")
```

## ২. ফর লুপ এবং রেঞ্জ (for loop & range)
একই কাজ বারবার করার জন্য বা কোনো লিস্ট/সিকোয়েন্সের প্রতিটি উপাদানের ওপর দিয়ে যাওয়ার জন্য `for` লুপ ব্যবহার করা হয়।

**`range()` ফাংশন:**
এটি একটি নির্দিষ্ট পরিমাণ সংখ্যার সিকোয়েন্স তৈরি করে। যেমন: `range(5)` মানে ০ থেকে ৪ পর্যন্ত সংখ্যা।

```python
# for loop এবং range() এর ব্যবহার
print("১ থেকে ৫ পর্যন্ত প্রিন্ট করা:")
for i in range(1, 6):  # ১ থেকে শুরু হবে, ৬ এর আগে (অর্থাৎ ৫ পর্যন্ত) থামবে
    print("বর্তমান সংখ্যা:", i)

print("\nML Club প্রিন্ট করা ৩ বার:")
for i in range(3):
    print("Machine Learning Club")
```

## ৩. হোয়াইল লুপ (while loop)
`while` লুপ ততক্ষণ পর্যন্ত চলতে থাকে, যতক্ষণ পর্যন্ত এর ভেতরের শর্তটি (condition) সত্যি বা `True` থাকে। শর্ত মিথ্যে হয়ে গেলে লুপ থেমে যায়।

```python
# while loop এর উদাহরণ
count = 1

print("While লুপ শুরু:")
while count <= 5:
    print("কাউন্ট:", count)
    count = count + 1  # অথবা count += 1 (এটি না দিলে লুপ অনন্তকাল চলতে থাকবে)
print(count)
```

## ৪. লুপ কন্ট্রোল: break এবং continue
- `break`: কোনো বিশেষ শর্ত মিললে লুপকে পুরোপুরি থামিয়ে দেওয়ার জন্য।
- `continue`: লুপের বর্তমান ধাপটি বাদ দিয়ে (skip করে) পরের ধাপে চলে যাওয়ার জন্য।

```python
print("--- break এর উদাহরণ ---")
for i in range(1, 10):
    if i == 5:
        print("৫ পাওয়া গেছে, লুপ ব্রেক করা হলো!")
        break
    print(i)

print("\n--- continue এর উদাহরণ ---")
for i in range(1, 6):
    if i == 3:
        continue  # ৩ এর জন্য নিচের কোডটি স্কিপ করবে
    print(i)
```

## ৫. নেস্টেড লুপ (Nested Loops)
একটি লুপের ভেতরে যখন আরেকটি লুপ থাকে, তখন তাকে নেস্টেড লুপ বলে। বাইরের লুপের প্রতিটি ধাপের জন্য ভেতরের লুপটি সম্পূর্ণভাবে রান করে।

```python
# নেস্টেড লুপের সাহায্যে একটি ছোট প্যাটার্ন তৈরি
print("স্টার প্যাটার্ন:")
for i in range(1, 4):         # বাইরের লুপ (সারি বা row)
    for j in range(1, 4):     # ভেতরের লুপ (কলাম বা column)
        print("*", end=" ")   # end=" " দিলে নতুন লাইনে যাবে না
    print()                   # ভেতরের লুপ শেষ হলে নতুন লাইনে যাওয়ার জন্য
```
