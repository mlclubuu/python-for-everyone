## ১. পাইথন কী? (Definition of Python)

- **Interpreted:** পাইথন কোড লাইন-বাই-লাইন এক্সিকিউট বা রান হয়, তাই আলাদা করে কম্পাইল করার প্রয়োজন হয় না।

## ২. পাইথন ভেরিয়েবল (Python Variables)
ভেরিয়েবল হলো একটি **কন্টেইনার** বা পাত্র, যেখানে আমরা কোনো ডেটা বা মান (value) সেভ করে রাখি যাতে পরে তা ব্যবহার করা যায়।

**ভেরিয়েবল নামকরণের নিয়ম:**
- নাম অবশ্যই লেটার (a-z, A-Z) বা আন্ডারস্কোর (`_`) দিয়ে শুরু হতে হবে।
- নম্বর দিয়ে শুরু করা যাবে না (যেমন: `1name` হবে না, `name1` হবে)।
- পাইথন কেস-সেনসিটিভ (যেমন: `age` এবং `Age` দুটি আলাদা ভেরিয়েবল হিসেবে কাজ করবে)।

```python
# ভেরিয়েবল তৈরি করা (Variable Assignment)
student_name = "Rahim"
student_id = 101
student_name = 12
# ভেরিয়েবল ব্যবহার করে আউটপুট দেখানো
print(student_name)
print(student_id)
```

## ৩. পাইথন ডেটা টাইপ (Python Data Types)
ভেরিয়েবলের মধ্যে আমরা কী ধরনের ডেটা রাখছি, তা নির্ধারণ করে ডেটা টাইপ। শুরুতে আমরা প্রধান ৪টি ডেটা টাইপ শিখব:

১. `int` (Integer): পূর্ণ সংখ্যা (যেমন: 10, -5, 100)

২. `float`: দশমিক সংখ্যা (যেমন: 3.14, 2.5, 0.99)

৩. `str` (String): টেক্সট বা লেখা, যা সিঙ্গল বা ডাবল কোটেশনের ভেতরে থাকে (যেমন: "Hello", 'Python')

৪. `bool` (Boolean): সত্য বা মিথ্যা (`True` অথবা `False`)

```python
# বিভিন্ন ডেটা টাইপের ভেরিয়েবল
age = 25              # int (পূর্ণ সংখ্যা)
gpa = 3.85            # float (দশমিক সংখ্যা)
course = "Backend"    # str (স্ট্রিং)
is_passed = True      # bool (বুলিয়ান)

# type() ফাংশন ব্যবহার করে ডেটা টাইপ চেক করা
print(type(age))
print(type(gpa))
print(type(course))
print(type(is_passed))
```

## ৪. গাণিতিক অপারেটর (Mathematical Operators)
পাইথনে গাণিতিক হিসাব করার জন্য নিচের অপারেটরগুলো ব্যবহার করা হয়:
- `+` : যোগ (Addition)
- `-` : বিয়োগ (Subtraction)
- `*` : গুণ (Multiplication)
- `/` : ভাগ (Division - এর আউটপুট সবসময় float বা দশমিক হয়)
- `%` : মডুলাস (Modulus - ভাগশেষ বের করার জন্য)
- `**` : এক্সপোনেনসিয়েশন (Exponentiation - পাওয়ার বা সূচক নির্ণয়)

```python
x = 15
y = 4

print("যোগফল:", x + y)
print("বিয়োগফল:", x - y)
print("গুণফল:", x * y)
print("ভাগফল:", x // y)
print("ভাগশেষ (Modulus):", x % y)
print("পাওয়ার (15 এর পাওয়ার 4):", x ** y)
```

## ৫. ইনপুট এবং প্রিন্ট (Input and Print Functions)
- `print()`: স্ক্রিনে কোনো আউটপুট বা রেজাল্ট দেখানোর জন্য ব্যবহার করা হয়।
- `input()`: ইউজারের কাছ থেকে ডায়নামিক ভাবে কোনো ডেটা বা ইনপুট নেওয়ার জন্য ব্যবহার করা হয়।

**গুরুত্বপূর্ণ নোট:** `input()` ফাংশন দিয়ে ইউজার যা-ই দেয়, পাইথন সেটাকে **String (str)** হিসেবে ধরে নেয়। তাই গাণিতিক কাজ করার জন্য সেটাকে `int()` বা `float()` দিয়ে পরিবর্তন করে নিতে হয়। এই প্রক্রিয়াকে **Type Casting** বলে।

```python
# স্ট্রিং ফরম্যাটিং (f-string)
name = "Sagar"
print(f"Hello, my name is {name}")

# ইউজার ইনপুট এবং টাইপ কাস্টিং (Type Casting)
number_input = input("যেকোনো একটি সংখ্যা লিখুন: ")
print("কাস্টিং এর আগে ডেটা টাইপ:", type(number_input))

# স্ট্রিং থেকে ফ্লোট-এ পরিবর্তন
converted_number = number_input
print("কাস্টিং এর পরে ডেটা টাইপ:", type(converted_number))
print("ওই সংখ্যার সাথে ১০ যোগ করলে হয়:", converted_number + "10")
```

## ৬. লাইভ প্রজেক্ট: সিম্পল ক্যালকুলেটর (Live Project: Simple Calculator)
এখন আমরা ওপরের সবগুলোর (Variable, Data Type, Operator, Input, Print) একসাথে ব্যবহার করে একটি সিম্পল ক্যালকুলেটর প্রজেক্ট তৈরি করব।

```python
print("=== প্র্যাকটিস ক্যালকুলেটরে স্বাগতম ===")

# ধাপ ১: ইউজারের কাছ থেকে ইনপুট নেওয়া
num1 = float(input("প্রথম সংখ্যাটি দিন: "))
num2 = float(input("দ্বিতীয় সংখ্যাটি দিন: "))

# ধাপ ২: গাণিতিক কাজ (Processing)
jogfol = num1 + num2
biyogfol = num1 - num2
gunfol = num1 * num2
vagfol = num1 / num2 if num2 != 0 else "Undefined (শূন্য দিয়ে ভাগ করা যায় না)"

# ধাপ ৩: আউটপুট দেখানো
print("\n--- ফলাফল ---")
print(f"প্রথম সংখ্যা: {num1} | দ্বিতীয় সংখ্যা: {num2}")
print(f"যোগফল: {jogfol}")
print(f"বিয়োগফল: {biyogfol}")
print(f"গুণফল: {gunfol}")
print(f"ভাগফল: {vagfol}")
```
